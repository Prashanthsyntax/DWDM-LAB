1️⃣ Euclidean Distance with Graph

```python
import numpy as np
import matplotlib.pyplot as plt

# Define points
A = np.array([2, 3])
B = np.array([5, 7])

# Euclidean Distance
distance = np.sqrt(np.sum((A - B) ** 2))
print("Euclidean Distance:", distance)

# Plot
plt.scatter(A[0], A[1], color='blue', label='Point A')
plt.scatter(B[0], B[1], color='red', label='Point B')

# Draw straight line
plt.plot([A[0], B[0]], [A[1], B[1]], color='black', linestyle='--')

plt.title("Euclidean Distance")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid()

plt.show()
```

2️⃣ Manhattan Distance with Graph

```python
import numpy as np
import matplotlib.pyplot as plt

# Define points
A = np.array([2, 3])
B = np.array([5, 7])

# Manhattan Distance
distance = np.sum(np.abs(A - B))
print("Manhattan Distance:", distance)

# Plot
plt.scatter(A[0], A[1], color='blue', label='Point A')
plt.scatter(B[0], B[1], color='red', label='Point B')

# L-shaped path
plt.plot([A[0], B[0]], [A[1], A[1]], color='green')
plt.plot([B[0], B[0]], [A[1], B[1]], color='green')

plt.title("Manhattan Distance")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid()

plt.show()
```

3️⃣ Minkowski Distance with Graph

```python
import numpy as np
import matplotlib.pyplot as plt

# Define points
A = np.array([2, 3])
B = np.array([5, 7])

# Minkowski Distance
p = 3
distance = np.sum(np.abs(A - B) ** p) ** (1/p)
print("Minkowski Distance:", distance)

# Plot
plt.scatter(A[0], A[1], color='blue', label='Point A')
plt.scatter(B[0], B[1], color='red', label='Point B')

plt.plot([A[0], B[0]], [A[1], B[1]], color='purple', linestyle='--')

plt.title("Minkowski Distance (p=3)")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid()

plt.show()
```
